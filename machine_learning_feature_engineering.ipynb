```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.metrics import classification_report, accuracy_score
```

```python
np.random.seed(42)

df = pd.DataFrame({
    'experience': np.random.randint(1, 20, 100),
    'age':        np.random.randint(22, 60, 100),
    'score':      np.random.randint(60, 100, 100),
    'department': np.random.choice(['HR', 'Tech', 'Finance'], 100),
    'salary':     np.random.randint(3000, 6000, 100),
    'promoted':   np.random.choice([0, 1], 100)
})
```

```python
df['salary_per_exp']= df['salary']/df['experience']
```

```python
df['age_group']= pd.cut(df['age'],
                        bins= [0, 35, 50, np.inf],
                        labels= ['young', 'mid', 'senior'])
```

```python
df_encoded= pd.get_dummies(df, columns=['department', 'age_group'], drop_first=True)
```

```python
X= df_encoded.drop('salary', axis=1)
y= df_encoded['salary']

X_train, X_test, y_train, y_test= train_test_split(X, y, random_state= 42, test_size=0.2)

scaler= StandardScaler()
X_train_scaled= scaler.fit_transform(X_train)
X_test_scaled= scaler.transform(X_test)

model= LinearRegression()
model.fit(X_train_scaled, y_train)
y_pred= model.predict(X_test_scaled)
rmse= np.sqrt(mean_squared_error(y_test, y_pred))
r2= r2_score(y_test, y_pred )

print('='*60)
print(f'Square root of mean squared error: {rmse:.2f}')
print(f'R squared value: {r2:.4f}')
print('='*60)

#the R2 value is negative, meaning our model is no better than just guessing the average predicted salaries. This is due to the fact that our values have no linear relationship since they were created at random
```

```python
from sklearn.tree import DecisionTreeRegressor

tree_model = DecisionTreeRegressor(random_state=42, max_depth=3)

tree_model.fit(X_train_scaled, y_train)

tree_pred = tree_model.predict(X_test_scaled)

tree_rmse = np.sqrt(mean_squared_error(y_test, tree_pred))
tree_r2 = r2_score(y_test, tree_pred)

print(" DECISION TREE RESULTS ")
print(f" RMSE: {tree_rmse:.2f}")
print(f" R² Score: {tree_r2:.4f}")

# I tried to use a DecissionTreeClassifier to see if the performace would improve, and it has.
```

```python
X2= df_encoded.drop('promoted', axis=1)
y2= df_encoded['promoted']

X2_train, X2_test, y2_train, y2_test= train_test_split(X2, y2, test_size=0.2, random_state=42)

scaler=StandardScaler()
X2_train_scaled= scaler.fit_transform(X2_train)
X2_test_scaled= scaler.transform(X2_test)

clf= LogisticRegression()
clf.fit(X2_train_scaled, y2_train)
y2_pred= clf.predict(X2_test_scaled)
print(classification_report(y2_test, y2_pred))
print('='*180)
print('When our model predicts not promoted, its 64% correct most of the time, and only 33% correct when predicting promoted(this can be dismissed as noise)')
print('Out of the 13 examples containing 0 (not promoted), our model managed to capture 54% of them, and only 33% of promoted, of 7 examples')
print('Your dataset is heavily imbalanced. Out of every 100 rows in your data, 95 of them belong to Class 0, and only 5 belong to Class 1 \nAccuracy is a highly deceptive metric because it treats all correct guesses as equally valuable, completely ignoring which class it got right')
print('='*180)
```
