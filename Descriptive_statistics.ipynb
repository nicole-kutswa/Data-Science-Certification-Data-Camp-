Descriptive statistics 

```python
import pandas as pd
import numpy as np
```

```python
salaries = [3200, 3400, 3100, 3600, 3300, 3250, 8500, 3150, 3400, 3300]
df = pd.DataFrame({'salary': salaries})
```

```python
salary_mean= df['salary'].mean()
salary_median= df['salary'].median()
salary_mode= df['salary'].mode()[0]
print(f"mean salary:{salary_mean}")
print(f"median salary:{salary_median}")
print(f"modal salary:{salary_mode}")

salary_max= df['salary'].max()
salary_min= df['salary'].min()
salary_range= salary_max- salary_min
print(f'salary range: {salary_range}')

salary_variance= df['salary'].var()
salary_std= df['salary'].std()
salary_skewness= df['salary'].skew()
print(f'salary std: {salary_std}')
print(f'salary variance: {salary_variance}')
print(f'salary skewness: {salary_skewness}')
```
