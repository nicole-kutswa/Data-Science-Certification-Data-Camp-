```python
import pandas as pd
import numpy as np
import re
import matplotlib.pyplot as plt
import seaborn as sns
```

```python


data = {
    'emp_id':     [1, 2, 2, 3, 4, 5, 6, 7, 8, 9],
    'name':       ['alice smith', 'BOB JONES', 'BOB JONES', 
                   'carol WHITE', '  dave brown  ', 'eve davis',
                   'frank miller', 'grace lee', 'henry CLARK', 'ian wright'],
    'email':      ['alice@company.com', 'bob@company', 'bob@company',
                   'carol@company.com', 'dave@company.com', 'not_an_email',
                   'frank@company.com', 'grace@company.com', 
                   'henry@company.com', 'ian@company.com'],
    'salary':     ['3200', '3400', '3400', '-500', '3300', 
                   '3250', '4500', '3150', '3400', '3800'],
    'hire_date':  ['2020-01-15', '2019-03-22', '2019-03-22', '2021-07-01',
                   '2018-11-30', '2022-05-14', '2017-08-09', '2023-01-20',
                   '2020-06-15', '2016-12-01'],
    'department': ['HR', 'Tech', 'Tech', 'Finance', 'HR',
                   'Finance', 'Tech', 'HR', 'Finance', 'Tech']
}

df = pd.DataFrame(data)
```

```python
duplicates= df['emp_id'].duplicated().sum()
print(duplicates)
#there is only one duplicate
df= df.drop_duplicates()
```

```python
df['name']= df['name'].str.strip().str.title()
df['name']
```

```python
df['salary']= pd.to_numeric(df['salary'], errors='coerce')
invalid_salary= (df['salary'] < 0)| (df['salary'].isna())
print(df.loc[invalid_salary])
```

```python
df= df.drop(df[invalid_salary].index)
```

```python
email_pattern= r'^[\w\.-]+@[\w\.-]+\.\w+$'

invalid_email= ~df['email'].str.match(email_pattern, na= True)
invalid_email_df= df[invalid_email]
print(invalid_email_df) 
```

```python
df['hire_date']= pd.to_datetime(df['hire_date'], format='%Y-%m-%d')
df['hire_year']= df['hire_date'].dt.year
df['days_employed']= (pd.Timestamp.now()-df['hire_date']).dt.days
```

```python
print("========================================")
print("       DATASET VALIDATION SUMMARY       ")
print("========================================\n")

# 1. Total rows remaining
print(f" Total Rows Remaining: {len(df)}")
print("-" * 40)

# 2. Remaining missing values
print(" Remaining Missing Values per Column:")
missing_counts = df.isna().sum()
if missing_counts.sum() == 0:
    print("   ✓ Perfect! No missing values detected.")
else:
    for col, count in missing_counts.items():
        if count > 0:
            print(f"   • {col}: {count} missing value(s)")
print("-" * 40)

# 3. All column dtypes
print(" Column Data Types:")
for col, dtype in df.dtypes.items():
    print(f"   • {col}: {dtype}")
print("-" * 40)

# 4. Confirm emp_id uniqueness
is_unique = df['emp_id'].is_unique
if is_unique:
    print(" Primary Key Validation: 'emp_id' is strictly UNIQUE.")
else:
    duplicate_count = df['emp_id'].duplicated().sum()
    print(f" WARNING: 'emp_id' has {duplicate_count} duplicated value(s)!")
print("\n========================================")
```
