Missingness And Correlation

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data = {
    'age':        [25, 32, 47, 51, 23, 38, 44, 29, None, 41],
    'salary':     [3200, 3400, 3100, 3600, 3300, 3250, 8500, 3150, 3400, None],
    'experience': [2, 5, 10, 15, 1, 7, None, 4, 6, 12],
    'score':      [88, 79, None, 91, 85, 76, 80, None, 83, 88]
}
df= pd.DataFrame(data)
df
```

Number and Percentage of Missing values

```python
total_size= df.size
missing_values= df.isnull().sum().sum()
percentage_missing= (missing_values/total_size)*100
print(percentage_missing)
```

```python
missing= df.isnull().sum()*100/len(df)
pct= missing.mean()
pct
```

```python
matrix= df.corr()
print(f'{matrix}')
plt.figure(figsize=(10, 8))
sns.heatmap(matrix, annot=True, cmap='coolwarm', fmt= '.2f')
plt.title('correlation heatmap')
plt.show()

print("Age and exprerience have a strong positive correlation, while salary and score have a strong negative correlation")
```
